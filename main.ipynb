```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import requests
import spacy
import re
```

### Testando se o modelo spacy foi carregado corretamente

Além disso, vamos declarar algumas funções essenciais

```python
nlp = spacy.load("pt_core_news_lg")
doc = nlp("Ouviram do Ipiranga às margens plácidas.")

def lemmetizar(string):
    string = re.sub("[^A-zÀ-ú0-9]+", ' ', string).lower()
    doc = nlp(string)
    return " ".join([token.lemma_ for token in doc if not token.is_stop])
```

```python
lemmetizar("O rato roeu a roupa do Rei de Roma")
```
